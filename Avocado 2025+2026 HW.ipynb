# Memo Figures Checking Point

```python
#Calculate avacado units sold for 2025 and 2026 
#Unit sold = "Tototal Bullk and Bags"

# Use only the Total U.S. rows. The file also has regions (West, Northeast...)
# and cities, so summing every row would count the same avocados more than once.
us2025 = data2025[data2025["Geography"] == "Total U.S."]
us2026 = data2026[data2026["Geography"] == "Total U.S."]

# Total units sold per year
units2025 = us2025["Total Bulk and Bags"].sum()
units2026 = us2026["Total Bulk and Bags"].sum()

print("Units Sold 2025 (full year, {} weeks): {:,.0f}".format(us2025["Period"].nunique(), units2025))
print("Units Sold 2026 (year to date, {} weeks): {:,.0f}".format(us2026["Period"].nunique(), units2026))

# Units sold by Type (Conventional vs Organic) for each year
by_type = pd.DataFrame({
    "2025": us2025.groupby("Type")["Total Bulk and Bags"].sum(),
    "2026": us2026.groupby("Type")["Total Bulk and Bags"].sum(),
})
print("\nUnits Sold by Type:")
print(by_type.apply(lambda col: col.map("{:,.0f}".format)))

# 2026 only has data through part of the year, so compare the same weeks of 2025
last_week_2026 = us2026["Period"].max()
units2025_same_weeks = us2025[us2025["Period"] <= last_week_2026]["Total Bulk and Bags"].sum()
change = (units2026 - units2025_same_weeks) / units2025_same_weeks * 100

print("\nSame period comparison (weeks 1-{}):".format(last_week_2026))
print("  2025: {:,.0f}".format(units2025_same_weeks))
print("  2026: {:,.0f}".format(units2026))
print("  Change: {:+.1f}%".format(change))
```

```python
# Implied retail dollars for each week = ASP x units sold
us2025["Retail Dollars"] = us2025["ASP Current Year"] * us2025["Total Bulk and Bags"]
us2026["Retail Dollars"] = us2026["ASP Current Year"] * us2026["Total Bulk and Bags"]

def summary(df):
    units = df["Total Bulk and Bags"].sum()
    dollars = df["Retail Dollars"].sum()
    # Average price per unit = total dollars / total units (weighted by volume)
    return units, dollars, dollars / units

units2025, dollars2025, avg2025 = summary(us2025)
units2026, dollars2026, avg2026 = summary(us2026)

print("2025 (full year): Retail Dollars ${:,.0f}  |  Avg Price per Unit ${:.2f}".format(dollars2025, avg2025))
print("2026 (year to date): Retail Dollars ${:,.0f}  |  Avg Price per Unit ${:.2f}".format(dollars2026, avg2026))

# Breakdown by Type (Conventional vs Organic)
print("\nBy Type:")
for year, df in [("2025", us2025), ("2026", us2026)]:
    by_type = df.groupby("Type")[["Total Bulk and Bags", "Retail Dollars"]].sum()
    by_type["Avg Price per Unit"] = by_type["Retail Dollars"] / by_type["Total Bulk and Bags"]
    for t, row in by_type.iterrows():
        print("  {} {:<12} Retail Dollars ${:,.0f}  |  Avg Price per Unit ${:.2f}".format(
            year, t, row["Retail Dollars"], row["Avg Price per Unit"]))

# 2026 only covers part of the year, so compare the same weeks of 2025
last_week_2026 = us2026["Period"].max()
units_same, dollars_same, avg_same = summary(us2025[us2025["Period"] <= last_week_2026])

print("\nSame period comparison (weeks 1-{}):".format(last_week_2026))
print("  Retail Dollars: 2025 ${:,.0f}  ->  2026 ${:,.0f}  ({:+.1f}%)".format(
    dollars_same, dollars2026, (dollars2026 - dollars_same) / dollars_same * 100))
print("  Avg Price per Unit: 2025 ${:.2f}  ->  2026 ${:.2f}  ({:+.1f}%)".format(
    avg_same, avg2026, (avg2026 - avg_same) / avg_same * 100))
```

```python
# Price Trends 2025 vs 2026
# Uses us2025, us2026 and the "Retail Dollars" column created in the earlier cells

colors = {"2025": "#2a78d6", "2026": "#eb6834"}

# Weekly average price per unit (both types combined) = retail dollars / units
def weekly_price(df):
    weekly = df.groupby("Period")[["Retail Dollars", "Total Bulk and Bags"]].sum()
    return weekly["Retail Dollars"] / weekly["Total Bulk and Bags"]

weekly2025 = weekly_price(us2025)
weekly2026 = weekly_price(us2026)

# Line chart: weekly price, 2025 vs 2026
fig, ax = plt.subplots(figsize=(12, 6))
for year, series in [("2025", weekly2025), ("2026", weekly2026)]:
    ax.plot(series.index, series.values, color=colors[year], linewidth=2, label=year)
    ax.text(series.index[-1] + 0.5, series.values[-1], year, color=colors[year], va="center", fontweight="bold")
ax.set_title("Weekly Average Price per Unit, Total U.S.")
ax.set_xlabel("Week of Year")
ax.set_ylabel("Avg Price per Unit ($)")
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: "${:.2f}".format(v)))
ax.grid(axis="y", alpha=0.3)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False)
plt.show()

# Line charts by Type: one chart for Conventional, one for Organic
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)
for ax, avocado_type in zip(axes, ["Conventional", "Organic"]):
    for year, df in [("2025", us2025), ("2026", us2026)]:
        series = weekly_price(df[df["Type"] == avocado_type])
        ax.plot(series.index, series.values, color=colors[year], linewidth=2, label=year)
    ax.set_title(avocado_type)
    ax.set_xlabel("Week of Year")
    ax.grid(axis="y", alpha=0.3)
    ax.spines[["top", "right"]].set_visible(False)
axes[0].set_ylabel("Avg Price per Unit ($)")
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: "${:.2f}".format(v)))
axes[0].legend(frameon=False)
plt.show()

# Monthly average price table with year-over-year change
def monthly_price(df):
    month = pd.to_datetime(df["Current Year Week Ending"]).dt.month
    monthly = df.groupby(month)[["Retail Dollars", "Total Bulk and Bags"]].sum()
    return monthly["Retail Dollars"] / monthly["Total Bulk and Bags"]

monthly = pd.DataFrame({"2025": monthly_price(us2025), "2026": monthly_price(us2026)})
monthly["Change %"] = (monthly["2026"] - monthly["2025"]) / monthly["2025"] * 100
monthly.index = pd.to_datetime(monthly.index, format="%m").strftime("%b")
monthly.index.name = "Month"
print("Monthly Average Price per Unit:")
print(monthly.round(2).to_string(na_rep="-"))

# Highest and lowest price week for each year
print("\nHighest and lowest price weeks:")
for year, series, df in [("2025", weekly2025, us2025), ("2026", weekly2026, us2026)]:
    dates = df.groupby("Period")["Current Year Week Ending"].first()
    print("  {}  High ${:.2f} (week ending {})  |  Low ${:.2f} (week ending {})".format(
        year, series.max(), dates[series.idxmax()], series.min(), dates[series.idxmin()]))
```

```python
# Volume Trends and Regional Growth: Conventional vs Organic

# Weekly units sold (Total U.S.), 2025 vs 2026
weekly_units2025 = us2025.groupby("Period")["Total Bulk and Bags"].sum()
weekly_units2026 = us2026.groupby("Period")["Total Bulk and Bags"].sum()

fig, ax = plt.subplots(figsize=(12, 6))
for year, series in [("2025", weekly_units2025), ("2026", weekly_units2026)]:
    ax.plot(series.index, series.values / 1e6, color=colors[year], linewidth=2, label=year)
    ax.text(series.index[-1] + 0.5, series.values[-1] / 1e6, year, color=colors[year], va="center", fontweight="bold")
ax.set_title("Weekly Units Sold, Total U.S.")
ax.set_xlabel("Week of Year")
ax.set_ylabel("Units Sold (millions)")
ax.grid(axis="y", alpha=0.3)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False)
plt.show()

# Organic share of units sold, by week
def organic_share(df):
    weekly = df.pivot_table(index="Period", columns="Type", values="Total Bulk and Bags", aggfunc="sum")
    return weekly["Organic"] / weekly.sum(axis=1) * 100

share2025 = organic_share(us2025)
share2026 = organic_share(us2026)
print("Organic share of units, weeks 1-{}: 2025 avg {:.1f}%  |  2026 avg {:.1f}%".format(
    last_week_2026, share2025[share2025.index <= last_week_2026].mean(), share2026.mean()))

# Growth by region and type, same weeks of each year (2026 is only a partial year)
regions = ["California", "West", "Plains", "South Central", "Midsouth", "Great Lakes", "Northeast", "Southeast"]

def region_units(df):
    df = df[(df["Geography"].isin(regions)) & (df["Period"] <= last_week_2026)]
    return df.pivot_table(index="Geography", columns="Type", values="Total Bulk and Bags", aggfunc="sum")

units_r2025 = region_units(data2025)
units_r2026 = region_units(data2026)
growth = (units_r2026 - units_r2025) / units_r2025 * 100
growth = growth.reindex(regions)

print("\nUnit growth by region, weeks 1-{} (2026 vs 2025):".format(last_week_2026))
print(growth.round(1).astype(str).add("%").to_string())

# Grouped bar chart: Conventional vs Organic growth by region
type_colors = {"Conventional": "#1baf7a", "Organic": "#4a3aa7"}
bar_height = 0.38
positions = range(len(regions))

fig, ax = plt.subplots(figsize=(11, 7))
for i, avocado_type in enumerate(["Conventional", "Organic"]):
    y = [p + (i - 0.5) * (bar_height + 0.02) for p in positions]
    bars = ax.barh(y, growth[avocado_type], height=bar_height, color=type_colors[avocado_type], label=avocado_type)
    ax.bar_label(bars, labels=["{:+.1f}%".format(v) for v in growth[avocado_type]], padding=4, fontsize=9)
ax.axvline(0, color="gray", linewidth=1)
ax.set_xlim(growth.min().min() - 6, growth.max().max() + 6)  # room for the % labels
ax.set_yticks(list(positions))
ax.set_yticklabels(regions)
ax.invert_yaxis()
ax.set_title("Unit Growth by Region, Weeks 1-{} (2026 vs 2025)".format(last_week_2026))
ax.set_xlabel("Growth in Units Sold (%)")
ax.grid(axis="x", alpha=0.3)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.08), ncol=2)
plt.show()
```

```python
# Average price per unit by region, same weeks of each year, and the change
def region_price(df):
    df = df[(df["Geography"].isin(regions)) & (df["Period"] <= last_week_2026)]
    dollars = (df["ASP Current Year"] * df["Total Bulk and Bags"]).groupby(df["Geography"]).sum()
    return dollars / df.groupby("Geography")["Total Bulk and Bags"].sum()

region_prices = pd.DataFrame({"2025": region_price(data2025), "2026": region_price(data2026)}).reindex(regions)
region_prices["Change $"] = region_prices["2026"] - region_prices["2025"]
region_prices["Change %"] = region_prices["Change $"] / region_prices["2025"] * 100

print("\nAverage price per unit by region, weeks 1-{}:".format(last_week_2026))
print(region_prices.round(2).to_string())
```
